```python
import numpy as np

# Q.1 Write a program to create a NumPy 1D-array with 5 elements and perform basic operations
arr_q1 = np.array([10, 20, 30, 40, 50])

print("Original Array:", arr_q1)

# a) Addition of 2 in all the elements
print("a) Add 2:", arr_q1 + 2)

# b) Multiply 3 with all the elements
print("b) Multiply by 3:", arr_q1 * 3)

# c) Divide every element by 2
print("c) Divide by 2:", arr_q1 / 2)
```

```python
# Q.2 Questions on Basic NumPy Array:

# a) Reverse the NumPy array
arr = np.array([1, 2, 3, 6, 4, 5])
reversed_arr = arr[::-1]
print("a) Reversed Array:", reversed_arr)

# b) Find the most frequent value and their indice(s)
def find_most_frequent(array):
    # Find unique values and their counts
    vals, counts = np.unique(array, return_counts=True)
    # Get the value with the maximum count
    most_freq_val = vals[np.argmax(counts)]
    # Find all indices where this value occurs
    indices = np.where(array == most_freq_val)[0]
    return most_freq_val, indices

# i. x array
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
freq_val_x, indices_x = find_most_frequent(x)
print(f"\nb) i. Array x -> Most frequent value: {freq_val_x}, Indices: {indices_x}")

# ii. y array
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
freq_val_y, indices_y = find_most_frequent(y)
print(f"b) ii. Array y -> Most frequent value: {freq_val_y}, Indices: {indices_y}")
```

```python
# Q.4 1-D NumPy array named <<Your Name>> with evenly spaced 25 numbers from 10 to 100
# Replacing <<Your Name>> with student_name
student_name = np.linspace(10, 100, 25)

print("Array:\n", student_name)
print("\nDimensions (ndim):", student_name.ndim)
print("Shape:", student_name.shape)
print("Total elements (size):", student_name.size)
print("Data type (dtype):", student_name.dtype)
print("Total number of bytes consumed (nbytes):", student_name.nbytes)

# Find the transpose using reshape() attribute
# Reshaping a 1D array of shape (25,) to a 2D column vector of shape (25, 1)
transposed_reshape = student_name.reshape(25, 1)
print("\nTranspose using reshape():\n", transposed_reshape)

# Can we do the same with T attribute?
print("\nCan we do the same with the .T attribute?")
print(f"Shape using .T: {student_name.T.shape}")
print("Answer: No. Because 'student_name' is a strictly 1-D array, the .T attribute does not change its shape. It must be at least a 2-D array (e.g., shape (1, 25)) for .T to transpose it into a column vector.")
```

```python
# Q.5 Create a 2-D Array of three rows and four columns, named ucs420_<<your_name>>
values = [10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]

# Replacing <<your_name>> with 'student'
ucs420_student = np.array(values).reshape(3, 4)
print("Original ucs420_student Array (3x4):\n", ucs420_student)

# Compute mean, median, max, min, unique elements
print("\nMean:", np.mean(ucs420_student))
print("Median:", np.median(ucs420_student))
print("Max:", np.max(ucs420_student))
print("Min:", np.min(ucs420_student))
print("Unique elements:", np.unique(ucs420_student))

# Reshape the array to four rows and three columns
reshaped_ucs420_student = ucs420_student.reshape(4, 3)
print("\nReshaped (4x3):\n", reshaped_ucs420_student)

# Resize the array to two rows and three columns 
# np.resize truncates or repeats values if the size changes (12 elements -> 6 elements)
resized_ucs420_student = np.resize(ucs420_student, (2, 3))
print("\nResized (2x3):\n", resized_ucs420_student)
```
