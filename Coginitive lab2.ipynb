```python
rollno = "1024170256" 

L = [int(digit) * 10 for digit in rollno]

print("i. Original List L:", L)

L.append(110) 
# Explanation: append(110) adds the element 110 to the very end of the list.
print("ii. After append(110):", L)

L.insert(5, 55) 
# Explanation: insert(5, 55) places the element 55 exactly at index 5, shifting subsequent elements to the right.
print("ii. After insert(5, 55):", L)

L.remove(110)
print("iii. After remove(110):", L)

popped_value = L.pop()
print(f"iii. After pop() (removed {popped_value}):", L)

L.sort()
print("iv. Sorted ascending:", L)

L.sort(reverse=True)
print("iv. Sorted descending:", L)

print("v. First three elements:", L[:3])
print("v. Last three elements:", L[-3:])

average = sum(L) / len(L)
L_greater_than_avg = [x for x in L if x > average]
print(f"vi. Elements greater than the average ({average}):", L_greater_than_avg)
```

```python
original_L = [int(digit) * 10 for digit in rollno]
scores = tuple(original_L[:8])
print("Scores tuple:", scores)

highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print(f"i. Highest score is {highest} at index {highest_index}. Lowest score is {lowest}, appearing {lowest_count} time(s).")

reversed_list = list(reversed(scores))
# Explanation: Tuples cannot be reversed in place because they are immutable (unchangeable after creation).
print("ii. Reversed tuple as list:", reversed_list)

search_val = input("Enter a score to search for: ")
if search_val.isdigit():
    search_val = int(search_val)
    if search_val in scores:
        print(f"iii. The first occurrence of {search_val} is at index {scores.index(search_val)}")
    else:
        print(f"iii. {search_val} is not present in the tuple.")

try:
    scores[0] = 100
except TypeError as e:
    print(f"iv. Error captured: {e}")
    # Explanation: This happens because tuples are immutable and do not support item assignment, unlike lists which are mutable.

first, second, *remaining = scores
print(f"v. Unpacked -> First: {first}, Second: {second}, Remaining: {remaining}")
```

```python
import random
from collections import Counter

roll_number_int = int(rollno)

random.seed(roll_number_int)

random_list = [random.randint(100, 900) for _ in range(100)]
print(f"i. random numbers: [{random_list}]")
odds = [x for x in random_list if x % 2 != 0]
print(f"ii. There are {len(odds)} odd numbers in the list.")

evens = [x for x in random_list if x % 2 == 0]
print(f"iii. There are {len(evens)} even numbers in the list.")

def is_prime(n):
    if n < 2: return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0: return False
    return True

prime_list = [x for x in random_list if is_prime(x)]
print(f"iv. There are {len(prime_list)} prime numbers in the list.")
print(f"    The prime numbers are: {prime_list}")

freq = Counter(random_list)
most_common_num, most_common_count = freq.most_common(1)[0]
print(f"v. The most frequent number is {most_common_num}, occurring {most_common_count} times.")
```

```python
my_dict = {
    "name": "Lokesh Sharma", 
    "roll_no": "1024170256", 
    "branch": "Computer Science", 
    "age": 20, 
    "city": "Patiala" 
}
print("Original Dictionary:", my_dict)

my_dict["location"] = my_dict.pop("city")
print("i. After renaming 'city' to 'location':", my_dict)

my_dict["cgpa"] = 8.75 

my_dict["age"] += 1

print("After steps ii & iii:", my_dict)

dict_copy1 = my_dict.copy()
popped_branch = dict_copy1.pop("branch")

dict_copy2 = my_dict.copy()
del dict_copy2["branch"]
# Explanation: pop() deletes the key and returns its value (useful if you need it), whereas del simply deletes the key-value pair and returns nothing.

my_dict = dict_copy2

print("\nv. Iterating Dictionary:")
for key, value in my_dict.items():
    print(f"{key} -> {value}")

print("\nvi. Safe fallback for missing key:")
if "email" in my_dict:
    print(my_dict["email"])
else:
    print("Safe fallback: The 'email' key does not exist in this dictionary.")

friend_dict = {
    "name": "Jane Doe",
    "roll_no": "0987654321",
    "branch": "Electrical Engineering",
    "age": 21,
    "location": "Chandigarh"
}

merged_dict = {**my_dict, **friend_dict}
print("\nvii. Merged Dictionary:", merged_dict)
# Explanation: When both dictionaries share a key, the values from the second dictionary (friend_dict) "win" and overwrite the first.

string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("\nviii. Dictionary with only string values:", string_only_dict)
```
