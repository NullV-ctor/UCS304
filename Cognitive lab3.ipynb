```python
# Create the sample dataset
data2 = {
    'Employee_ID': [101, 102, 103, 104, 105],
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Edward'],
    'Department': ['HR', 'IT', 'IT', 'Marketing', 'Sales'],
    'Age': [29, 34, 41, 28, 38],
    'Salary': [50000, 70000, 65000, 55000, 60000],
    'Years_of_Experience': [4, 8, 10, 3, 12],
    'Joining_Date': ['2020-03-15', '2017-07-19', '2013-06-01', '2021-02-10', '2010-11-25'],
    'Gender': ['Female', 'Male', 'Male', 'Female', 'Male'],
    'Bonus': [5000, 7000, 6000, 4500, 5000],
    'Rating': [4.5, 4.0, 3.8, 4.7, 3.5]
}

pd.DataFrame(data2).to_csv('employees.csv', index=False)
df_emp = pd.read_csv('employees.csv')

# a) Shape (number of rows and columns)
print(f"a) Shape of DataFrame: {df_emp.shape}\n")

# b) Summary (data types and non-null counts)
print("b) Summary of DataFrame:")
df_emp.info()
print("\n")

# c) Generate descriptive statistics
print("c) Descriptive Statistics:")
display(df_emp.describe())

# d) First 5 rows and last 3 rows
print("\nd) First 5 rows:")
display(df_emp.head())
print("\nLast 3 rows:")
display(df_emp.tail(3))

# e) Calculate specific statistics
print("\ne) Calculated Statistics:")
print(f"  i. Average salary: {df_emp['Salary'].mean()}")
print(f" ii. Total bonus: {df_emp['Bonus'].sum()}")
print(f"iii. Youngest employee's age: {df_emp['Age'].min()}")
print(f" iv. Highest performance rating: {df_emp['Rating'].max()}")

# f) Sort by Salary in descending order
print("\nf) Sorted by Salary (Descending):")
display(df_emp.sort_values(by='Salary', ascending=False))

# g) Add Performance Category column
conditions = [
    (df_emp['Rating'] >= 4.5),
    (df_emp['Rating'] >= 4.0) & (df_emp['Rating'] < 4.5),
    (df_emp['Rating'] < 4.0)
]
choices = ['Excellent', 'Good', 'Average']
df_emp['Performance_Category'] = np.select(conditions, choices)

# h) Identify missing values
print("\nh) Missing values in each column:")
print(df_emp.isnull().sum())

# i) Rename Employee_ID to ID
df_emp.rename(columns={'Employee_ID': 'ID'}, inplace=True)

# j) Find employees with >5 years experience AND in IT department
print("\nj) Employees with >5 years experience AND in IT:")
target_employees = df_emp[(df_emp['Years_of_Experience'] > 5) & (df_emp['Department'] == 'IT')]
display(target_employees)

# k) Add Tax column (10% of Salary)
df_emp['Tax'] = df_emp['Salary'] * 0.10

# Display final modified DataFrame before saving
print("\nFinal Modified DataFrame (with new columns):")
display(df_emp)

# l) Save modified DataFrame to a new CSV file
df_emp.to_csv('employees_modified.csv', index=False)
print("\nl) Modified DataFrame successfully saved to 'employees_modified.csv'")
```
