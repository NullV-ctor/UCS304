```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C to every reading using vectorization
temp_added = temperature + 2
print("a. Added 2°C:", temp_added)

# b. Convert all temperatures from Celsius to Fahrenheit
temp_f = (9/5) * temperature + 32
print("b. Fahrenheit:", temp_f)

# c. Identify all readings greater than 32°C using Boolean indexing
high_temps = temperature[temperature > 32]
print("c. Readings > 32°C:", high_temps)

# d. Count how many readings exceed 32°C
high_temps_count = np.sum(temperature > 32)
print("d. Count > 32°C:", high_temps_count)

# e. Explanation of efficiency
print("\ne. Vectorization and Boolean indexing allow operations to execute across entire arrays at the C-level, bypassing slow Python for-loops. This makes processing faster and code more concise.")
```

```python
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Calculate the total steps recorded
print("a. Total steps:", np.sum(steps))

# b. Calculate the mean number of steps
print("b. Mean steps:", np.mean(steps))

# c. Find the maximum and minimum values
print(f"c. Max steps: {np.max(steps)}, Min steps: {np.min(steps)}")

# d. Calculate the total steps for each day (columns, axis=0)
print("d. Total per day (axis=0):", np.sum(steps, axis=0))

# e. Calculate the total steps for each user (rows, axis=1)
print("e. Total per user (axis=1):", np.sum(steps, axis=1))

# f. Find the user/day position containing the maximum number of steps
max_pos = np.unravel_index(np.argmax(steps), steps.shape)
print("f. Max steps position (User row, Day column):", max_pos)
```

```python
import numpy as np

# a. Create original array
original = np.array([1, 2, 3, 4, 5, 6])
print("a. Original:", original)

# b & c. Slice and modify (Memory View)
subset = original[1:5]
subset[0] = 999
print("\nb & c. Subset modified (999):", subset)
print("   Original array is AFFECTED:", original)

# d. Slice using .copy() and modify
original = np.array([1, 2, 3, 4, 5, 6]) # Reset original
copied_subset = original[1:5].copy()
copied_subset[0] = 500
print("\nd. Copied subset modified (500):", copied_subset)
print("   Original array is UNAFFECTED:", original)

# e. Create 1 to 12 array and reshape to 3x4
matrix = np.arange(1, 13).reshape(3, 4)
print("\ne. 3x4 Matrix:\n", matrix)

# f. Extract specific elements
print("\nf. First row:", matrix[0])
print("   Last row:", matrix[-1])
print("   Second column:", matrix[:, 1])
print("   Rows 1-2 & Cols 2-3:\n", matrix[1:3, 2:4])

# g, h, & i. Flatten vs Ravel
flat_arr = matrix.flatten()
ravel_arr = matrix.ravel()

ravel_arr[0] = 99
print("\nh. Modifying ravel() AFFECTS original matrix:\n", matrix)

flat_arr[1] = 88
print("\ni. Modifying flatten() does NOT affect original matrix:\n", matrix)

# j. Array properties
print(f"\nj. Shape: {matrix.shape}, ndim: {matrix.ndim}, size: {matrix.size}, dtype: {matrix.dtype}")
```

```python
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])

# a. Display shape and dimensions
print("a. Shape of X:", X.shape, ", Dimensions:", X.ndim)

# b. Calculate transpose
X_T = X.T
print("\nb. Transpose (X.T):\n", X_T)
print("   The transpose swaps rows and columns, turning user profiles into feature rows.")

# c. Calculate matrix product
XTX = X_T @ X
print("\nc. Matrix product (X.T @ X):\n", XTX)

# d. Calculate matrix inverse
XTX_inv = np.linalg.inv(XTX)
print("\nd. Matrix inverse (X.T @ X)^-1:\n", XTX_inv)

# e. Implement Ordinary Least Squares equation
beta = XTX_inv @ X_T @ y
print("\ne. OLS Coefficients (beta):", beta)

# f. Explain coefficients
print("\nf. The coefficients represent the individual weights of Sleep, Activity, and Stress. They indicate how much the assistance score changes per unit increase in each feature.")

# g. Predict assistance score for new user
new_user = np.array([5, 40, 7])
prediction = new_user @ beta
print("\ng. Predicted assistance score for new user:", prediction)
```
